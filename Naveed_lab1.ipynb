```python
def world_tour_agent(budget, continent, days):
    if continent == "Europe":
        cost_per_day = 20000
        city = "Paris"
    elif continent == "Asia":
        cost_per_day = 12000
        city = "Tokyo"
    elif continent == "Africa":
        cost_per_day = 10000
        city = "Cairo"
    elif continent == "North America":
        cost_per_day = 28000
        city = "New York"
    elif continent == "South America":
        cost_per_day = 16000
        city = "Rio de Janeiro"
    elif continent == "Oceania":
        cost_per_day = 24000
        city = "Sydney"
    else:
        print("Continent not found.")
        return

    total_cost = cost_per_day * days

    if total_cost <= budget:
        print("Recommended City:", city)
        print("Continent:", continent)
        print("Days:", days)
        print("Total Cost: Rs.", total_cost)
        print("Status: Trip approved within budget!")
    else:
        print("Trip to", city, "exceeds your budget.")
        print("Required: Rs.", total_cost)
        print("Your Budget: Rs.", budget)


budget = int(input("Enter your budget: "))
continent = input("Enter continent (Europe/Asia/Africa/North America/South America/Oceania): ")
days = int(input("Enter number of days: "))

world_tour_agent(budget, continent, days)
```
