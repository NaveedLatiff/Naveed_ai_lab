```python
from google import genai

client = genai.Client(api_key="")

MODEL = "models/gemini-3.5-flash"

def html_generator(user_input):

    prompt = f"""
You are an HTML website generator.

Task: Create a simple, clean, single-file HTML5 website based on the user request.
Rules:
1. Output ONLY HTML code. No explanation, no markdown,
2. And simple css

User request: {user_input}

Generate the HTML code now:
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text

print("Gemini HTML Generator Started (type 'exit' to stop)\n")

while True:
    user = input("You: ")

    if user.lower() == "exit":
        print("Goodbye ")
        break

    try:
        reply = html_generator(user)
        print(reply)
    except Exception as e:
        print("Error:", e)
```
