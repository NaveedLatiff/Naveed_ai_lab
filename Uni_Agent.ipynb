```python
from google import genai

client = genai.Client(api_key="")

MODEL = "models/gemini-3.5-flash"

FUUAST_FACULTY_DATA = {
     "section_6a": {
        "AI": "Miss Uzma",
        "NLP": "Sir Khalid",
        "N&SC": "Sir Kashif Riffat",
        "CC": "Miss Naheed"
    },
    "section_6b": {
        "AI": "Miss Uzma",
        "NLP": "Sir Sarim",
        "N&SC": "Sir Akhtar Raza",
        "CC": "Miss Naheed"
    },
     "section_6c": {
        "AI": "Miss Uzma",
        "NLP": "Sir khalid",
        "N&SC": "Sir Kashif Riffat",
        "CC": "Miss Naheed"
    }
}

def fuuast_agent(user_input):

    prompt = f"""
You are an AI assistant for Federal Urdu University of Arts, Sciences & Technology (FUUAST).

Task: Provide accurate, clear, and helpful information about FUUAST based on the user's inquiry.
if u some user ask about details like who is teaching AI to students of section 6B for that you can
use this dictionary => {FUUAST_FACULTY_DATA} and if the information is not on the website or in the dictionary i provided to you
then reply user that i dont have enough information about it



User query: {user_input}

Answer:
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text

print("FUUAST Information Agent Started (type 'exit' to stop)\n")

while True:
    user = input("You: ")

    if user.lower() == "exit":
        print("Goodbye!")
        break

    try:
        reply = fuuast_agent(user)
        print("\nAgent:\n", reply, "\n")
    except Exception as e:
        print("Error:", e)
```
